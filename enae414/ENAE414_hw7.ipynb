## Uniform Flow with Two Sources

Consider a two-dimensional, steady, incompressible, and irrotational flow consisting of:

- Uniform freestream velocity $v_\infty$ in the positive $x$-direction.
- A source of strength $\Lambda$ at $(-h,0)$.
- A source of strength $\Lambda$ at $(h,0)$.

By the principle of superposition,

$$
\boxed{
\phi = \phi_\infty + \phi_{s_1} + \phi_{s_2}
}
$$

For incompressible, irrotational flow, the velocity potential satisfies Laplace's equation away from the sources:

$$
\boxed{\nabla^2\phi=0}
$$

In Cartesian coordinates,

$$
\frac{\partial^2\phi}{\partial x^2}
+
\frac{\partial^2\phi}{\partial y^2}=0.
$$

In polar coordinates,

$$
\frac{1}{r}\frac{\partial}{\partial r}
\left(r\frac{\partial\phi}{\partial r}\right)
+
\frac{1}{r^2}\frac{\partial^2\phi}{\partial\theta^2}
=0.
$$

## Part (a): Velocity Potential — Separation of Variables

Assume a separable solution to Laplace's equation:

$$
\phi(r,\theta)=R(r)\Theta(\theta).
$$

Substituting into Laplace's equation gives

$$
\frac{1}{r}\frac{d}{dr}
\left(rR'\Theta\right)
+
\frac{1}{r^2}R\Theta''=0.
$$

Multiplying by $r^2/(R\Theta)$,

$$
r^2\frac{R''}{R}
+
r\frac{R'}{R}
+
\frac{\Theta''}{\Theta}=0.
$$

Introduce the separation constant $n^2$:

$$
\boxed{
r^2R''+rR'-n^2R=0
}
$$

$$
\boxed{
\Theta''+n^2\Theta=0
}
$$

### Radial solution

Assume a power-law solution:

$$
R(r)=r^m.
$$

Then

$$
R'=mr^{m-1},
\qquad
R''=m(m-1)r^{m-2}.
$$

Substitution gives

$$
m(m-1)r^m+mr^m-n^2r^m=0.
$$

Therefore,

$$
m^2-n^2=0
\quad\Longrightarrow\quad
m=\pm n.
$$

The radial solution is

$$
\boxed{
R(r)=A_nr^n+B_nr^{-n}
}
$$

for $n\geq1$.

### Angular solution

The angular differential equation is

$$
\Theta''=-n^2\Theta.
$$

Therefore,

$$
\boxed{
\Theta(\theta)=C_n\cos(n\theta)+D_n\sin(n\theta)
}
$$

### Special case: $n=0$

For $n=0$, the radial equation becomes

$$
r^2R''+rR'=0.
$$

Rearranging,

$$
\frac{d}{dr}(rR')=0.
$$

Integrating,

$$
rR'=A_0,
$$

$$
R'=\frac{A_0}{r}.
$$

Integrating again,

$$
\boxed{
R(r)=A_0\ln r+B_0
}
$$

The logarithmic term is particularly important because it describes the
velocity potential of a two-dimensional source.

### General solution

For a single-valued potential in a full angular domain, the separated solution can be written as

$$
\boxed{
\begin{aligned}
\phi(r,\theta)
={}&A_0\ln r+B_0\\
&+\sum_{n=1}^{\infty}
\left(A_nr^n+B_nr^{-n}\right)
\left[C_n\cos(n\theta)+D_n\sin(n\theta)\right].
\end{aligned}
}
$$

The boundary conditions and flow configuration determine which terms are required.

## Part (a): Velocity Potential in Cartesian Coordinates

For a two-dimensional source of strength $\Lambda$,

$$
\phi_s(r)=\frac{\Lambda}{2\pi}\ln r.
$$

The two sources are located at $(-h,0)$ and $(h,0)$.

Their respective radial distances are

$$
r_1=\sqrt{(x+h)^2+y^2},
$$

$$
r_2=\sqrt{(x-h)^2+y^2}.
$$

Therefore,

$$
\phi_{s_1}
=
\frac{\Lambda}{2\pi}
\ln\sqrt{(x+h)^2+y^2},
$$

$$
\phi_{s_2}
=
\frac{\Lambda}{2\pi}
\ln\sqrt{(x-h)^2+y^2}.
$$

Using the logarithmic identity

$$
\ln\sqrt{A}=\frac12\ln A,
$$

we obtain

$$
\phi_{s_1}
=
\frac{\Lambda}{4\pi}
\ln\left[(x+h)^2+y^2\right],
$$

$$
\phi_{s_2}
=
\frac{\Lambda}{4\pi}
\ln\left[(x-h)^2+y^2\right].
$$

For the uniform freestream,

$$
\phi_\infty=v_\infty x.
$$

Applying superposition,

$$
\boxed{
\begin{aligned}
\phi(x,y)
={}&v_\infty x\\
&+\frac{\Lambda}{4\pi}
\ln\left[(x+h)^2+y^2\right]\\
&+\frac{\Lambda}{4\pi}
\ln\left[(x-h)^2+y^2\right].
\end{aligned}
}
$$

An arbitrary additive constant has been omitted because it does not affect
the velocity field:

$$
\mathbf{V}=\nabla\phi.
$$

## Part (b): Stream Function

For two-dimensional incompressible flow, the stream function is defined by

$$
u=\frac{\partial\psi}{\partial y},
\qquad
v=-\frac{\partial\psi}{\partial x}.
$$

The stream function for a uniform flow in the positive $x$-direction is

$$
\psi_\infty=v_\infty y.
$$

For a two-dimensional source,

$$
\psi_s=\frac{\Lambda}{2\pi}\theta.
$$

The angular coordinates relative to each source are

$$
\theta_1=\operatorname{atan2}(y,x+h),
$$

$$
\theta_2=\operatorname{atan2}(y,x-h).
$$

Applying superposition,

$$
\psi=\psi_\infty+\psi_{s_1}+\psi_{s_2}.
$$

Therefore,

$$
\boxed{
\begin{aligned}
\psi(x,y)
={}&v_\infty y\\
&+\frac{\Lambda}{2\pi}\operatorname{atan2}(y,x+h)\\
&+\frac{\Lambda}{2\pi}\operatorname{atan2}(y,x-h).
\end{aligned}
}
$$

The function $\operatorname{atan2}(y,x)$ is used instead of
$\tan^{-1}(y/x)$ to preserve the correct quadrant.

## Part (c): Velocity Field

The velocity field is obtained from the gradient of the velocity potential:

$$
\boxed{
\mathbf{V}=\nabla\phi
}
$$

In Cartesian coordinates,

$$
\mathbf{V}
=
\frac{\partial\phi}{\partial x}\hat{\mathbf{x}}
+
\frac{\partial\phi}{\partial y}\hat{\mathbf{y}}.
$$

### Horizontal velocity

$$
u=\frac{\partial\phi}{\partial x}.
$$

Differentiating the velocity potential,

$$
\boxed{
\begin{aligned}
u(x,y)
={}&v_\infty\\
&+\frac{\Lambda}{2\pi}
\frac{x+h}{(x+h)^2+y^2}\\
&+\frac{\Lambda}{2\pi}
\frac{x-h}{(x-h)^2+y^2}.
\end{aligned}
}
$$

### Vertical velocity

$$
v=\frac{\partial\phi}{\partial y}.
$$

Therefore,

\boxed{
\begin{aligned}
v(x,y)
={}&\frac{\Lambda}{2\pi}

\frac{y}{(x-h)^2+y^2}.

}

The resulting velocity field is

$$

## Part (d): Stagnation Points

Stagnation points occur where the velocity vanishes:

$$
\boxed{
u(x,y)=0,
\qquad
v(x,y)=0.
}
$$

For two positive sources, the vertical velocity is

$$

\left[

\frac{\Lambda}{2\pi}

\frac{1}{x+h}+\frac{1}{x-h}

$$

$$
\frac{1}{x+h}+\frac{1}{x-h}
=
\frac{2x}{x^2-h^2}.
$$

$$
v_\infty+\frac{\Lambda x}{\pi(x^2-h^2)}=0.
$$

Multiplying by $\pi(x^2-h^2)$,

$$

## Part (f): Pressure Coefficient and Nondimensionalization

For steady, incompressible, inviscid, irrotational flow, Bernoulli's equation gives

$$
p+\frac12\rho V^2
=
p_\infty+\frac12\rho v_\infty^2.
$$

Rearranging,

$$
p-p_\infty
=
\frac12\rho(v_\infty^2-V^2).
$$

The pressure coefficient is defined as

$$
C_p=
\frac{p-p_\infty}{\frac12\rho v_\infty^2}.
$$

Substitution gives

$$
\boxed{
C_p=1-\frac{V^2}{v_\infty^2}
}
$$

Since

$$
V^2=u^2+v^2,
$$

the final expression is

$$
\boxed{
C_p(x,y)
=
1-\frac{u(x,y)^2+v(x,y)^2}{v_\infty^2}.
}
$$

### Nondimensional variables

Define

$$
X=\frac{x}{h},
\qquad
Y=\frac{y}{h}.
$$

The nondimensional velocity potential and stream function are

$$
\boxed{
\phi^*=\frac{\phi}{v_\infty h},
\qquad
\psi^*=\frac{\psi}{v_\infty h}.
}
$$

The nondimensional velocity components are

$$
\boxed{
u^*=\frac{u}{v_\infty},
\qquad
v^*=\frac{v}{v_\infty}.
}
$$

Therefore,

$$
\boxed{
C_p=1-(u^{*2}+v^{*2}).
}
$$

$$

$$
\boxed{

}

```python
def potential(x, y, paras):
    v_free, lamb, h = paras
    phi = v_free * x + (lamb/(4*np.pi)) * np.log((x+h)**2 + y**2) + (lamb/(4*np.pi)) * np.log((x-h)**2 + y**2)
    return phi
def stream(x, y, paras):
    v_free, lamb, h = paras
    psi = v_free * y + (lamb/(2*np.pi)) * np.atan2(y, (x-h)) + (lamb/(2*np.pi)) * np.atan2(y, (x+h))
    return psi
def velocity(x, y, paras):
    v_free, lamb, h = paras
    u = v_free + (lamb/(2*np.pi)) * ((x+h)/((x+h)**2 + y**2)) + (lamb/(2*np.pi)) * ((x-h)/((x-h)**2 + y**2))
    v = (lamb/(2*np.pi)) * (y/((x+h)**2 + y**2)) + (lamb/(2*np.pi)) * (y/((x-h)**2 + y**2))
    return u, v
def cp(x, y, paras):
    v_free, lamb, h = paras
    u, v = velocity(x, y, paras)
    cp = 1 - (1/v_free)**2 * (u**2 + v**2)
    return cp
def stagnation_xy(function, paras):
    x, y = sp.symbols("x y", real=True)
    u, v = function(x, y, paras)
    # Find stagnation points along the x-axis
    u_axis = sp.cancel(u.subs(y, 0))
    # Solve numerator = 0
    numerator = sp.fraction(u_axis)[0]
    solutions = sp.nroots(numerator)
    points = []
    for sol in solutions:
        if abs(float(sp.im(sol))) < 1e-10:
            x0 = float(sp.re(sol))
            # Exclude singularities
            if all(abs(x0 - s) > 1e-8 for s in [0, paras[2], 2*paras[2]]):
                points.append((x0, 0.0))
    return points

d = .25
x_h = np.linspace(-3-d, 3+d, 650)
y_h = np.linspace(-3-d, 3+d, 650)
X_h, Y_h = np.meshgrid(x_h, y_h)
paras = 1, 1, 1
v_free, lamb, h = paras
phi = potential(X_h, Y_h, paras)/(v_free * h)
psi = stream(X_h, Y_h, paras)/(v_free * h)
u, v = velocity(X_h, Y_h, paras)
u = u / v_free
v = v / v_free
stag_points = stagnation_xy(velocity, paras)
x_stag = [point[0] for point in stag_points]
y_stag = [point[1] for point in stag_points]
Cp = cp(X_h, Y_h, paras)

plt.figure(figsize=(9, 7))
# Pressure coefficient
cp_plot = plt.contourf(X_h, Y_h, Cp, levels=np.linspace(-1, 1, 100), cmap="coolwarm", extend="both", alpha=0.7)
# Stream function
plt.contour(X_h, Y_h, psi, levels=20, colors="blue", linewidths=1, linestyles="solid")
# Dividing streamlines
plt.contour(X_h, Y_h, psi, levels=[-1, 1],colors="purple",linewidths=3, linestyles="solid") 
# Velocity potential
plt.contour(X_h, Y_h, phi, levels=20, colors="black", linestyles="solid", linewidths=1)
# Stagnation points
plt.scatter(x_stag, y_stag, color="red", zorder=5)
# Sources
plt.scatter([-1, 1], [0, 0], color="black", zorder=5)
# Formatting
plt.colorbar(cp_plot, label=r"$C_p$")
plt.xlabel(r"$\frac{x}{h}$")
plt.ylabel(r"$\frac{y}{h}$")
plt.title("Uniform Flow with Two Sources")
plt.xlim(-3, 3)
plt.ylim(-3, 3)
# Legend
plt.plot([], [], color="blue", linewidth=2, label=r"Stream Function: $\psi$")
plt.plot([], [], color="black", linewidth=2, label=r"Velocity Potential: $\phi$")
plt.plot([], [], color="purple", linewidth=2, label=r"Dividing streamlines")
plt.scatter([], [], color="red", label="Stagnation Points")
plt.scatter([], [], color="black", label="Sources")
plt.legend(loc="upper left",bbox_to_anchor=(1.25, 1),fontsize=10)
plt.show()
```

```python
def potential2(x, y, paras):
    v_free, lamb, h, a = paras
    phi_free = v_free * x
    phi_s1 = (lamb/(4*np.pi)) * np.log(x**2 + y**2)
    phi_s2 = (-lamb/(8*np.pi)) * np.log((x-h)**2 + y**2)
    phi_s3 = (-a*lamb/(4*np.pi)) * np.log((x-2*h)**2 + y**2)
    phi = phi_free + phi_s1 + phi_s2 + phi_s3
    return phi
def stream2(x, y, paras):
    v_free, lamb, h, a = paras
    psi_free = v_free * y
    psi_s1 = (lamb/(2*np.pi)) * np.atan2(y, x)
    psi_s2 = (-lamb/(4*np.pi)) * np.atan2(y, (x-h))
    psi_s3 = (-lamb * a /(2*np.pi)) * np.atan2(y, (x-2*h))
    psi = psi_free + psi_s1 + psi_s2 + psi_s3
    return psi
def velocity2(x, y, paras):
    v_free, lamb, h, a = paras
    u = v_free + (lamb/(2*np.pi))*(x/(x**2 + y**2)) - (lamb/(4*np.pi))*((x-h)/((x-h)**2+y**2)) - (a*lamb/(2*np.pi))*((x-2*h)/((x-2*h)**2+y**2))
    v = (lamb/(2*np.pi))*(y/(x**2 + y**2)) - (lamb/(4*np.pi))*(y/((x-h)**2+y**2)) - (a*lamb/(2*np.pi))*(y/((x-2*h)**2+y**2))
    return u, v
def cop2(x, y, paras):
    v_free, lamb, h, a = paras
    u = velocity2(x, y, paras)[0]
    v = velocity2(x, y, paras)[1] 
    cp = 1 - (u**2 + v**2)/v_free**2
    return cp
x = np.linspace(-3.25, 3.25, 650)
y = np.linspace(-3.25, 3.25, 650)
X, Y = np.meshgrid(x, y)
paras = 1, 1, 1, 0.5
v_free, lamb, h, a = paras
equipot = potential2(X, Y, paras)
streamlines = stream2(X, Y, paras)
CoP = cop2(X, Y, paras)
stag_points = stagnation_xy(velocity2, paras)
x_stag2 = [point[0] for point in stag_points]
y_stag2 = [point[1] for point in stag_points]


#plotting
plt.figure(figsize=(9, 7))
# Pressure coefficient
cp_plot2 = plt.contourf(X, Y, CoP, levels=np.linspace(-1, 1, 100), cmap="coolwarm", extend="both", alpha=0.7, linestyles = "solid")
plt.contour(X, Y, equipot, levels = 20, colors = "black", linestyles = "solid", linewidths = 1)
plt.contour(X, Y, streamlines, levels = 20, colors = "blue", linestyles = "solid", linewidths = 1)
plt.contour(X_h, Y_h, streamlines, levels=[0],colors="purple",linewidths=3, linestyles="solid") 
plt.scatter([0, h, 2*h], [0, 0, 0], color="black", zorder=5)
plt.scatter(x_stag2, y_stag2, color="red", zorder=5)
# Formatting
plt.colorbar(cp_plot2, label=r"$C_p$")
plt.xlabel(r"$\frac{x}{h}$")
plt.ylabel(r"$\frac{y}{h}$")
plt.title("Uniform Flow with Two Sources")
plt.xlim(-3, 3)
plt.ylim(-3, 3)
# Legend
plt.plot([], [], color="blue", linewidth=2, label=r"Stream Function: $\psi$")
plt.plot([], [], color="black", linewidth=2, label=r"Velocity Potential: $\phi$")
plt.plot([], [], color="purple", linewidth=2, label=r"Dividing streamlines")
plt.scatter([], [], color="red", label="Stagnation Points")
plt.scatter([], [], color="black", label="Sources")
plt.legend(loc="upper left",bbox_to_anchor=(1.25, 1),fontsize=10)
plt.show()
```
